# Qiskit Guide: Making Circuits

This notebook introduces the basics of building quantum circuits with Qiskit.

By the end of this guide, you will know how to:

- create a quantum circuit
- add qubits and classical bits
- apply simple gates such as `H`, `X`, and `CX`
- measure the circuit
- draw the circuit so you can see what you built

If you are brand new to quantum computing, think of a circuit as a list of instructions that tells Qiskit what should happen to each qubit.

## 1. Import Qiskit

Start by importing the main circuit class. This is the object that will hold your quantum instructions.

````python
from qiskit import QuantumCircuit
````

## 2. Create a Simple Circuit

The code below creates a circuit with:

- `2` qubits, which hold the quantum information
- `2` classical bits, which store measurement results

You can think of classical bits as the place where the final answer gets written down.

````python
qc = QuantumCircuit(2, 2)
qc
````

## 3. Add Gates

Now we add some basic gates:

- `h(0)` puts qubit 0 into superposition
- `cx(0, 1)` uses qubit 0 to control qubit 1
- `x(1)` flips qubit 1

The numbers tell Qiskit which qubit each gate should act on.

````python
qc.h(0)
qc.cx(0, 1)
qc.x(1)

qc.draw("text")
````

## 4. Measure the Circuit

A measurement turns the qubit state into a classical result.

The line `measure([0, 1], [0, 1])` means:

- measure qubit 0 into classical bit 0
- measure qubit 1 into classical bit 1

````python
qc.measure([0, 1], [0, 1])

qc.draw("text")
````

## 5. Build a Fresh Example From Scratch

Creating a new circuit is often easier than reusing an old one while you are learning. The example below makes a Bell-state-style circuit, which is one of the most common beginner examples in Qiskit.

````python
bell = QuantumCircuit(2, 2)
bell.h(0)
bell.cx(0, 1)
bell.measure([0, 1], [0, 1])

bell.draw("text")
````

## 6. Useful Pattern To Remember

A simple workflow for making circuits in Qiskit is:

1. create a `QuantumCircuit`
2. add gates
3. measure qubits if you want classical output
4. draw or run the circuit

Template:

```python
qc = QuantumCircuit(number_of_qubits, number_of_classical_bits)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
qc.draw("text")
```

Try changing the circuit by adding more gates such as `z`, `y`, or additional `x` gates to see how the diagram changes.
